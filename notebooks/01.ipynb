## (a) How does the Basel II Accord's emphasis on risk measurement influence the need for interpretability in credit scoring models?

The Basel II Capital Accord establishes international regulatory standards for bank capital adequacy and places strong emphasis on risk measurement, transparency, documentation, and model governance. Under Basel II, banks using Internal Ratings-Based (IRB) approaches must ensure that their credit risk models are transparent, auditable, and defensible to regulators.

This requirement creates a strong need for model interpretability in credit scoring systems. Regulators, auditors, and risk management teams must be able to understand why a customer receives a particular credit risk score. If a model functions as a “black box,” its decisions cannot be easily explained or justified, creating compliance and audit challenges.

Basel II also requires extensive documentation of model assumptions, data sources, feature definitions, validation procedures, and monitoring strategies. Every component of the model, including target variable construction and feature engineering decisions, must be clearly documented and justified.

In addition, Basel II requires ongoing model validation and performance monitoring. Models must be regularly assessed for stability, accuracy, and drift over time. This is easier to achieve when using interpretable and well-understood models.

Therefore, while highly complex models may offer greater predictive power, financial institutions often prioritize interpretable approaches because they support regulatory compliance, transparency, auditability, and stakeholder trust.

## (b) Why is creating a proxy variable necessary in this project, and what are the potential business risks of using such a proxy?
Why a Proxy Variable is Necessary

The Xente transaction dataset does not contain a direct default or loan repayment label, which is normally required for supervised credit risk modeling. Since the eCommerce platform has not historically provided credit products, there is no existing record indicating which customers repaid loans and which customers defaulted.

As a result, a proxy target variable must be created to approximate credit risk. The project uses RFM (Recency, Frequency, Monetary) analysis combined with clustering techniques to identify customers who exhibit low engagement, infrequent purchases, low transaction values, and long periods of inactivity. These customers are treated as higher-risk proxy cases.

The underlying assumption is that customers demonstrating weaker transactional behavior are more likely to present higher credit risk than active, high-value customers.

Potential Business Risks
1. Label Noise

The proxy may incorrectly classify customers. Some creditworthy customers may be labeled as high risk, while genuinely risky customers may be labeled as low risk. This can reduce model accuracy and lead to poor lending decisions.

2. Concept Drift

The relationship between RFM behavior and actual credit default may change over time. Behavioral patterns that indicate risk today may not accurately predict future repayment behavior.

3. Regulatory Scrutiny

Regulators may question whether the proxy truly represents default risk. Without empirical evidence linking the proxy to actual loan outcomes, regulatory approval and compliance may become more difficult.

4. Fairness and Bias

Behavioral patterns may unintentionally reflect demographic or socioeconomic differences. If not carefully monitored, the proxy could introduce biased lending decisions and fairness concerns.

5. Business Impact

An overly conservative proxy may classify too many customers as risky, resulting in excessive loan rejections, lower customer acquisition, reduced product adoption, and lost revenue opportunities.

Conclusion

The proxy variable is a practical solution when real default data is unavailable; however, it remains a modeling assumption rather than actual ground truth. As real repayment data becomes available, the model should be retrained and validated using true default outcomes.

## (c) What are the trade-offs between using a simple, interpretable model such as Logistic Regression with Weight of Evidence (WoE) and a complex model such as Gradient Boosting for credit scoring?

Credit scoring models often require balancing interpretability and predictive performance.

Logistic Regression with WoE
Advantages
Highly interpretable and easy to explain.
Model coefficients clearly show how each feature affects risk.
WoE transformation produces monotonic and stable relationships between variables and risk.
Easier to document, validate, and defend to regulators.
Supports adverse-action explanations and reason codes for rejected applicants.
Fast to deploy and maintain.
Disadvantages
Assumes mostly linear relationships between features and risk.
May fail to capture complex interactions and nonlinear patterns.
Often produces lower predictive accuracy compared to advanced machine learning models.
Gradient Boosting (e.g., XGBoost or LightGBM)
Advantages
Captures complex nonlinear relationships and feature interactions.
Generally achieves higher predictive performance.
Better at modeling heterogeneous customer behavior.
Can identify subtle risk patterns that simpler models may miss.
Disadvantages
Difficult to interpret and explain.
Requires additional explanation tools such as SHAP or LIME.
More challenging to validate and document for regulatory purposes.
Higher maintenance and monitoring requirements.
Harder to communicate model decisions to business stakeholders and regulators.
Recommended Approach

For Bati Bank's BNPL solution, a dual-model strategy is recommended:

Primary Model: Logistic Regression with WoE-transformed features should be used as the production credit scoring model because it satisfies Basel II interpretability and regulatory requirements.
Challenger Model: A Gradient Boosting model (such as XGBoost or LightGBM) should run alongside the primary model to evaluate potential performance improvements and capture additional predictive signals.

This approach provides regulatory compliance, transparency, and explainability while still allowing the bank to benefit from the predictive strength of advanced machine learning models.
